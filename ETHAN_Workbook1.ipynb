```python
# libaries loops and lists
import numpy as np
import pandas as pd 
# can use "as" to nickname the libraries
```

```python
# create an array and make an array of numbers 0 to 100 (exclusive)
random_list = np.arange(stop=100)
print(random_list)
```

```python
#A1 assignment: use np.arange to create a new list, called even_list_1 that contains all the numbers from 0 to 100 (exclusive)
even_list_1 = np.arange(step=2, stop=100)
print(even_list_1)
```

```python
#the [] allows us to access elements from an array
print(random_list[1])

# can use the : operator to access elements across a certain range
print(random_list[5:15])
# note that the last element (15) is exclusive
# can also access a number at the beginning/end of a list by using : to indicate we want python to go all the way to the beginning/end
print(random_list[5:])
print(random_list[:5])
```

```python
#A2
# assignment: print the element at the 90th index, print the elements from 10-21, print the last 10 elements
print(random_list[90])
print(random_list[10:22])
print(random_list[90:])

# try not to put numerical values in variable names
```

```python
# math ops on our list
# get the list to start at 1 instead of 0
random_list = random_list + 1
print(random_list)

#can also do it for logarithms
print(np.log(random_list))

```

```python
# boolean indexing can also be performed element-wise across an array
# boolean index array that shows where all the Falses are
print(random_list % 2 == 0)

even_list_2 = random_list[random_list % 2 ==0]
# this filters out the stuff that returns false, so only stuff that returns true will be added into even list 2

print(even_list_2)
```

```python
# initialize empty array
even_list_3 = np.array([])

for i in range(0,101):
    if i % 2 == 0:
        even_list_3 = np.append(arr=even_list_3, values = i)

print(even_list_3)
```

```python
# list comprehension

# make an array of numbers 0 -200 (exclusive)
big_list = np.arange(200)

# multiply the numbers resulting in a list
even_list_4 = [2*x for x in big_list if x <=50]
# so you can put all these terms inside the brackets
# [action for x in array if condition]
even_list_4 = np.array(even_list_4)
print(even_list_4)
```

```python
#A3
test_list = np.array([ -99.52287875, -100.        ,  -99.39794001, -100.        ,
        -99.30103   ,  -99.04575749,  -99.69897   ,  -99.22184875,
        -99.30103   ,  -99.52287875,  -99.30103   ,  -99.09691001,
        -99.04575749,  -99.15490196,  -99.04575749,  -99.52287875,
        -99.69897   ,  -99.52287875,  -99.09691001,  -99.39794001,
        -99.22184875,  -99.69897   ,  -99.22184875,  -99.39794001])
# add 100 to each number in the list
test_list = test_list +100
print(test_list)

# use np.power() to take 10 to the power of each value in test_list
test_list = np.power(10, test_list)
print(test_list)

# use np.round() to round each of the values in the list to the nearest whole numbers
test_list = np.round(test_list, 1)
print(test_list)

# print the first five elements
print(test_list[:5])
```

```python
# dataframes and pandas (i love pandas)

# turns the file into a variable
gapminder = pd.read_csv("gapminder.csv")

#opens the dataset
gapminder


```

```python
# this prints the data in the row at index number 1
print(gapminder.iloc[1])

# the index (the 0th column) is also integers, this also works with .loc
print(gapminder.loc[1])

# can also print the data in the first five rows
print(gapminder.head(5))
```

```python
# can also print using iloc
print(gapminder.iloc[0:5])

```

```python
# can also print specific columns
print(gapminder.loc[:,'continent'])
```

```python
# another way to print data in the column is to use the list index
print(gapminder['continent'])
```

```python
# can print multiple columns at once
print(gapminder[['continent', 'country']])
```

```python
# filtering dataframes
# say we want to filter the dates from 1980 to 2000
print((gapminder['year'] >= 1980))

filtered_gapminder = gapminder[(gapminder['year']>=1980) & (gapminder['year'] <= 2000)]
print(filtered_gapminder)

# make sure to use parentheses to keep stuff neat
# and comment
```
